```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Data structure : [restaurant_id,2021,2022,2023,2024]
sale_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 12000, 140000, 160000, 190000],   # Beijing Bites
    [3, 20000, 203000, 260000, 300000],   # Pizza Hub
    [4, 180000, 210000, 2400000, 270000], # Burger Point
    [5, 160000, 1850000, 2050000, 230000] # Chai Point
])

print("=== Zomato sale analysis ===")
print("Sales Data:\n", sale_data.shape)
print("\n Sample data for 1st 3 restaurant",sale_data[:, 1:])
```

Total sale per year

```python
print(np.sum(sale_data, axis=0))
yearly
_total = np.sum(sale_data[:, 1:], axis = 0)
print(yearly_total)
```

# minimum sale per restaurant

```python
min_sales = np.min(sale_data[:,1:] , axis=1)
print(min_sales)

# maximum sales per year

max_sales = np.max(sale_data[:,1:] axis = 0)
print(max_sales)

# Average sales per restaurant

avg_sales = np.mean(sale_data[:,1:], axis = 1)
print(avg_sales)
```

```python
cumcum = np.cumsum(sale_data[:,1:] , axis = 1)
print(cumcum)

plt.figure(figsize = (10,6))
plt.plot(np.mean(cumcum,axis=0))
plt.title("Average cumulative sales accross all resttaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector addition" , vector1 + vector2)
print("\n Multiplication vector" , vector1 * vector2)
print("\nDot Product",np.dot(vector1,vector2))

angle = np.arccos(np.dot(vector1,vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
